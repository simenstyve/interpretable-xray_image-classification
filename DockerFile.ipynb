Dockerfile (prêt pour CAMCIS / Azure / On-prem)
Compatible CPU/GPU (selon base image).
FROM python:3.10-slim

WORKDIR /app

COPY . /app

# OpenCV + pgmpy + ultralytics
RUN apt-get update && apt-get install -y \
        libgl1-mesa-glx \
        libglib2.0-0 \
    && pip install --no-cache-dir \
        fastapi \
        uvicorn[standard] \
        opencv-python \
        joblib \
        ultralytics \
        pgmpy \
        scikit-learn \
        numpy \
        pandas

EXPOSE 8000

CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
